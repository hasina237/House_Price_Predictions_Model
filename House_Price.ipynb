```python
df['Total_Area'] = df['Gr Liv Area'].fillna(0) + df['Total Bsmt SF'].fillna(0)

df['House_Age'] = (df['Yr Sold'] - df['Year Built']).clip(lower=0)

df['Total_Bathrooms'] = df['Full Bath'] + (0.5 * df['Half Bath'])

df['Total_Rooms'] = df['TotRms AbvGrd']  

df['Garage_Score'] = df['Garage Area'].fillna(0) * df['Garage Cars'].fillna(0)

df['Remod_age'] = (df['Yr Sold'] - df['Year Remod/Add']).clip(lower=0)

df['Is_Remodeled'] = (df['Year Remod/Add'] != df['Year Built']).astype(int)
```

```python
print("Before selection:", type(df))

cols = ['Gr Liv Area','Overall Qual','Garage Area','Full Bath','Half Bath','Bedroom AbvGr','Year Built',
        'Year Remod/Add','Total Bsmt SF','1st Flr SF','2nd Flr SF','Garage Cars','TotRms AbvGrd',
        'Lot Area','Overall Cond','Mas Vnr Area','Bsmt Full Bath','Open Porch SF','Fireplaces','Remod_age','Garage_Score',
        'Total_Rooms','Total_Bathrooms','House_Age','Total_Area','SalePrice']

df = df[cols]

print("After selection:", type(df))
```

```python
print("Right before isna:", type(df))
```

```python
df.fillna(df.median(numeric_only=True),inplace=True)
```

```python
df.isna().sum()
df['SalePrice'].value_counts()
```

```python
X = df.drop('SalePrice',axis=1)
y = df['SalePrice']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.fit_transform(X_test)

y_train_log = np.log1p(y_train)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train_scaled,y_train_log)
```

```python
y_pred_log = model.predict(X_test_scaled)
```

```python
y_pred = np.expm1(y_pred_log)
y_true = y_test
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

print('MAE',mean_absolute_error(y_true,y_pred))
print('R2', r2_score(y_true,y_pred))
print('MSE', mean_squared_error(y_true,y_pred))
```

```python
import joblib

joblib.dump(model, 'House_Price_Model.pkl')
```
